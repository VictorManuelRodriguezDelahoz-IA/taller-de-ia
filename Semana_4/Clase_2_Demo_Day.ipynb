````python
# Verificador de proyecto
import os
from pathlib import Path

def verificar_proyecto(ruta_proyecto: str) -> dict:
    """Verifica que el proyecto tenga todos los archivos necesarios."""
    
    archivos_requeridos = [
        "README.md",
        ".env.example",
        "requirements.txt",
        ".gitignore"
    ]
    
    carpetas_recomendadas = [
        "src",
        "tests",
        "data"
    ]
    
    proyecto_path = Path(ruta_proyecto)
    
    print("🔍 Verificando estructura del proyecto...\n")
    print("="*80)
    
    # Verificar archivos
    print("\n📄 Archivos Requeridos:")
    archivos_ok = 0
    for archivo in archivos_requeridos:
        existe = (proyecto_path / archivo).exists()
        status = "✅" if existe else "❌"
        print(f"  {status} {archivo}")
        if existe:
            archivos_ok += 1
    
    # Verificar carpetas
    print("\n📁 Carpetas Recomendadas:")
    carpetas_ok = 0
    for carpeta in carpetas_recomendadas:
        existe = (proyecto_path / carpeta).exists()
        status = "✅" if existe else "⚠️"
        print(f"  {status} {carpeta}/")
        if existe:
            carpetas_ok += 1
    
    # Calcular score
    total_items = len(archivos_requeridos) + len(carpetas_recomendadas)
    completado = archivos_ok + carpetas_ok
    score = completado / total_items
    
    print("\n" + "="*80)
    print(f"\nScore: {completado}/{total_items} ({score:.0%})")
    
    if score >= 0.8:
        print("\n✅ ¡Proyecto bien estructurado!")
    elif score >= 0.5:
        print("\n⚠️ Proyecto aceptable, pero puede mejorar")
    else:
        print("\n❌ Proyecto necesita más trabajo")
    
    return {
        "score": score,
        "archivos_ok": archivos_ok,
        "carpetas_ok": carpetas_ok
    }

# Ejemplo de uso
# verificar_proyecto("./mi_proyecto")
print("✅ Verificador de proyecto listo")
````

## 🎤 Parte 2: Estructura de Presentación

### Formato Recomendado (5-7 minutos)

#### 1. Introducción (1 min)
- **Problema**: ¿Qué problema resuelves?
- **Solución**: ¿Cómo lo resuelves con IA?
- **Valor**: ¿Por qué importa?

**Ejemplo**:
> "Muchos estudiantes tienen dificultad para encontrar información en documentos largos. 
> Creé un asistente RAG que permite hacer preguntas en lenguaje natural sobre cualquier documento. 
> Esto reduce el tiempo de búsqueda de 30 minutos a 30 segundos."

#### 2. Arquitectura (1-2 min)
- Diagrama simple del sistema
- Componentes principales
- Tecnologías usadas

**Incluye**:
- LLM usado (GPT-4, etc.)
- Vector store (ChromaDB, etc.)
- Framework (LangChain, LangGraph)
- Tools/APIs integradas

#### 3. Demo en Vivo (2-3 min)
- **Caso 1**: Caso de uso básico
- **Caso 2**: Caso más complejo
- **Caso 3**: Edge case o característica especial

**Tips**:
- Prepara los casos de antemano
- Ten un video backup por si falla
- Explica qué está pasando

#### 4. Resultados y Métricas (1 min)
- Precisión/Accuracy
- Tiempo de respuesta
- Casos de prueba pasados
- Feedback de usuarios

#### 5. Aprendizajes y Próximos Pasos (1 min)
- ¿Qué aprendiste?
- Desafíos enfrentados
- Mejoras futuras
- Roadmap

## 🎬 Parte 3: Tips para Demos Efectivos

### Antes del Demo

1. **Practica 3+ veces**
   - Cronometra tu presentación
   - Practica con amigos/familia
   - Grábate y revisa

2. **Prepara tu entorno**
   - Cierra tabs innecesarias
   - Aumenta tamaño de fuente
   - Prepara casos de prueba
   - Ten backup (video/screenshots)

3. **Verifica todo funcione**
   - API keys válidas
   - Internet estable
   - Dependencias instaladas
   - Datos de prueba listos

### Durante el Demo

1. **Empieza fuerte**
   - Hook en primeros 30 segundos
   - Problema claro y relatable

2. **Muestra, no digas**
   - Demo > Slides
   - Casos reales > Teoría

3. **Maneja errores con gracia**
   - Si algo falla, ten plan B
   - Explica qué debería pasar
   - Usa backup si es necesario

4. **Cuenta una historia**
   - Conecta emocionalmente
   - Usa ejemplos concretos
   - Muestra el impacto

### Después del Demo

1. **Q&A preparado**
   - Anticipa preguntas
   - Sé honesto sobre limitaciones
   - Comparte aprendizajes

2. **Call to Action**
   - GitHub repo
   - Demo link
   - Contacto

## 🏆 Parte 4: Preparación para Hackathons

### Antes del Hackathon

#### 1. Setup Técnico
```bash
# Template de proyecto
mkdir mi-hackathon-project
cd mi-hackathon-project

# Estructura básica
touch README.md .env.example requirements.txt
mkdir src tests data

# Git
git init
```

#### 2. Herramientas Listas
- ✅ LangChain/LangGraph instalado
- ✅ API keys configuradas
- ✅ Templates de código
- ✅ Datasets de prueba

#### 3. Equipo
- Define roles claros
- Establece comunicación
- Plan de trabajo

### Durante el Hackathon

#### Timeboxing (24h hackathon)

**Horas 0-2: Ideación**
- Brainstorm ideas
- Evaluar viabilidad
- Decidir proyecto
- Dividir tareas

**Horas 2-6: MVP**
- Implementar funcionalidad básica
- Probar que funciona
- No perfeccionar aún

**Horas 6-12: Features**
- Agregar funcionalidades
- Integrar componentes
- Testing básico

**Horas 12-18: Pulir**
- UI/UX
- Error handling
- Optimización

**Horas 18-22: Demo**
- Preparar presentación
- Crear video
- Documentar
- Practicar pitch

**Horas 22-24: Buffer**
- Últimos ajustes
- Backup de todo
- Descansar

### Tips de Hackathon

1. **Scope pequeño, ejecución grande**
   - Mejor un feature perfecto que 10 a medias

2. **Demo > Código**
   - Los jueces ven el demo, no el código
   - Invierte tiempo en presentación

3. **Usa lo que sabes**
   - No aprendas tecnología nueva
   - Usa templates y ejemplos

4. **Documenta mientras construyes**
   - README incremental
   - Screenshots/videos

5. **Duerme algo**
   - 4-6 horas mínimo
   - Mejor código con descanso
